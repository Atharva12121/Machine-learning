```python
import pandas as pd
import seaborn as sns #Seaborn is a Python library specifically designed for data visualization, built on top of Matplotlib
#Data Visualization is the graphical representation of data and information using visual elements like charts, graphs,
#maps, and diagrams. It helps people understand complex data patterns, trends, and relationships by presenting data in an easily
#interpretable, visual format.
from sklearn import linear_model
```

```python
df=pd.read_csv('age_premium_data.csv')
```

```python
df
```

```python
sns.lmplot(x='Age',y='Premium',data=df)
```

```python
# now by linearn regration the model hass to train
reg=linear_model.LinearRegression()
```

```python
reg.fit(df[['Age']].values,df[['Premium']].values)
```

```python
reg.predict([[21]]) # predict ate the age of 21
```
