```python
import pandas as pd
```

```python
dataset=pd.read_csv('data.csv')
```

```python
dataset
```

```python
x=dataset.iloc[:,1:4].values
```

```python
x
```

```python
y=dataset.Purchased.values
```

```python
y
```

```python
from sklearn.preprocessing import LabelEncoder # chatogrise data convert into numeric data
```

```python
le=LabelEncoder()
```

```python
x[:,0]=le.fit_transform(x[:,0])
```

```python
x
```

```python
# now spliting the data for training and testing
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
# now we create navie bayes model in this their are multiple class so we use GausianNG
from sklearn.naive_bayes import GaussianNB
```

```python
model=GaussianNB()
```

```python
model.fit(x_train,y_train)
```

```python
model.predict(x_test)
```

```python
x_test
```

```python
model.predict([[0,99,20000]])
```

```python
model.score(x_test,y_test)# 90% accurarcy
```
