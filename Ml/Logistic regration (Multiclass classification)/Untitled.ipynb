```python
import pandas as pd
```

```python
dataset=pd.read_csv('iris_data.csv')
```

```python
dataset
```

```python
# here  species in catogrial form show we reaplace in numeric value
dataset['Species'].unique()
```

```python
dataset['Species'].replace({'Iris-virginica': 1, 'Iris-setosa': 2, 'Iris-versicolor': 3}, inplace=True)
```

```python
dataset
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test=train_test_split(dataset[['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']],dataset['Species'],test_size=0.2)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr=LogisticRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
lr.predict(x_test)
```

```python
x_test
```

```python
lr.predict([[7.21, 4.1, 6.1,2.8]])
```

```python
# now we check the score
lr.score(x_test,y_test)
```

```python
# this show our model 100% accuracy 
```
