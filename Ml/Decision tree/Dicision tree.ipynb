```python
import pandas as pd
```

```python
data=pd.read_csv('dicision.csv')
```

```python
data
```

```python
# encode this into numeric form so we import
from sklearn.preprocessing import LabelEncoder
```

```python
outlook=LabelEncoder()
humidity=LabelEncoder()
windy=LabelEncoder()	
play=LabelEncoder()
```

```python
data['outlook']=outlook.fit_transform(data['outlook'])
data['humidity']=outlook.fit_transform(data['humidity'])
data['windy']=outlook.fit_transform(data['windy'])
data['play']=outlook.fit_transform(data['play'])
```

```python
data
```

```python
# seperate independent varaibel and dependent var
x=data.iloc[:,0:3]
```

```python
y=data.play
```

```python
x
```

```python
y
```

```python
# now split data for trainig and testing
from sklearn.model_selection import  train_test_split
```

```python
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.2)
```

```python
# now import ddecision tree classifier
from sklearn.tree import DecisionTreeClassifier
```

```python
#classifier=DecisionTreeClassifier(criterion='gini')
classifier=DecisionTreeClassifier(criterion='entropy')
```

```python
len(x_train)
```

```python
classifier.fit(x_train,y_train)
```

```python
classifier.predict(x_test)
```

```python
x_test
```

```python
classifier.score(x_test,y_test)
```
